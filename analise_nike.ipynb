```python
import pandas as pd

# 1. Carregar os dados das vendas da Nike
df = pd.read_csv('vendas_nike.csv')

# 2. Visualizar as primeiras 5 linhas
display(df.head())

# Informações gerais sobre colunas e tipos de dados
print("\n--- Informações do Dataset ---")
df.info()
```

```python
# --- Análise Exploratória de Dados (KPIs e Insights) ---

# 1. Métrica Principais
faturamento_total = df['Receita_Total'].sum()
total_unidades = df['Quantidade'].sum()
ticket_medio = df['Receita_Total'].mean()

print(f"💰 Faturamento Total: €{faturamento_total:,.2f}")
print(f"📦 Total de Produtos Vendidos: {total_unidades} unidades")
print(f"📊 Ticket Médio por Pedido: €{ticket_medio:,.2f}\n")

# 2. Vendas por Produto
vendas_produto = df.groupby('Produto')['Receita_Total'].sum().sort_values(ascending=False)
print("--- 👟 Faturamento por Produto ---")
print(vendas_produto)

# 3. Vendas por Região
vendas_regiao = df.groupby('Regiao')['Receita_Total'].sum().sort_values(ascending=False)
print("\n--- 🌍 Faturamento por Região ---")
print(vendas_regiao)
```

```python
import plotly.express as px

# 1. Gráfico de Faturamento por Produto
fig_produto = px.bar(
    df.groupby('Produto', as_index=False)['Receita_Total'].sum().sort_values(by='Receita_Total', ascending=True),
    x='Receita_Total',
    y='Produto',
    orientation='h',
    title='Faturamento Total por Produto (Nike)',
    labels={'Receita_Total': 'Faturamento (€)', 'Produto': 'Modelo'},
    color='Receita_Total',
    color_continuous_scale='Reds'
)
fig_produto.show()

# 2. Gráfico de Distribuição por Região
fig_regiao = px.pie(
    df,
    names='Regiao',
    values='Receita_Total',
    title='Distribuição de Vendas por Região',
    hole=0.4
)
fig_regiao.show()
```
